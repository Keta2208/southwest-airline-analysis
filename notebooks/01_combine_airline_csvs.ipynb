# Step 1: Combine the monthly BTS files

Merges the 12 monthly BTS On-Time Performance CSVs for 2024, keeps the columns needed for the analysis, and filters to Southwest (WN) and American Airlines (AA). The output (`combined_airline_data.csv`) is the data source for the Tableau workbook. Put the 12 downloaded CSVs in one folder and set `path` before running (the notebook was written in Google Colab, where they were uploaded to `/content/`).

```python
import pandas as pd
import glob
import os

# Path to our 12 CSV files uploaded manually
path = "/content/"

# Columns we want (may not exist in all files)
desired_cols = [
    "FlightDate",
    "Reporting_Airline",
    "IATA_CODE_Reporting_Airline",
    "Origin",
    "OriginCityName",
    "OriginStateName",
    "Dest",
    "DestCityName",
    "DestStateName",
    "DepTime",
    "DepDelay",
    "TaxiOut",
    "TaxiIn",
    "ArrTime",
    "ArrDelay",
    "Cancelled",
    "Diverted",
    "ActualElapsedTime",
    "CarrierDelay",
    "WeatherDelay",
    "NASDelay",
    "SecurityDelay",
    "LateAircraftDelay"
]

# Storing all monthly DataFrames before combining
Combining = []

# Getting all CSV files in /content
csv_files = glob.glob(os.path.join(path, "*.csv"))

for file in csv_files:
    print("Reading:", file)

    df = pd.read_csv(file, low_memory=False)

    # KEEP ONLY the columns we need
    existing_cols = [c for c in desired_cols if c in df.columns]
    df = df[existing_cols]

    # FILTERing ONLY WN and it's competitor AA since the file is too big
    df = df[df['Reporting_Airline'].isin(['WN', 'AA'])]

    # Adding filtered data
    Combining.append(df)


# Combining all months together into one large DataFrame
combined = pd.concat(Combining, ignore_index=True)

# Saving final output
output_path = os.path.join(path, "combined_airline_data.csv")
combined.to_csv(output_path, index=False)

print("DONE! Saved to:", output_path)
```
